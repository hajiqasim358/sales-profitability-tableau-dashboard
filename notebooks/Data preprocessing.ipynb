```python
# Setting the dataset path.
import pandas as pd

file_path = "/content/drive/MyDrive/Data_Visualization_Project/Superstore.csv"

df = pd.read_csv(file_path, encoding='latin1')
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
```

```python
df.info()
```

```python
## Creating new columns
# year column
df['Year'] = df['Order Date'].dt.year
```

```python
# Month Column, Gives Jan, Feb, Mar etc, useful for charts
df['Month'] = df['Order Date'].dt.strftime('%b')
```

```python
## Profit Ratio (VERY IMPORTANT)
# This is key for your story
# Shows how much profit per sale
df['Profit Ratio'] = df['Profit'] / df['Sales']
```

```python
# Create Profit Category (for storytelling)
# will Help highlight loss areas in dashboard
df['Profit Category'] = df['Profit'].apply(lambda x: 'Profit' if x > 0 else 'Loss')
```

```python
df.head()
```

```python
df[['Year','Month','Profit Ratio','Profit Category']].head()
```

```python
df.isnull().sum()
```

```python
## Profit Ratio
## Sales
## Profit

df[['Sales','Profit','Profit Ratio']].describe()
```

```python
df['Profit Ratio'] = df['Profit Ratio'].replace([float('inf'), -float('inf')], 0)
df['Profit Ratio'] = df['Profit Ratio'].fillna(0)
```

```python
df = df.drop(['Row ID', 'Postal Code', 'Customer ID', 'Product ID'], axis=1)
```
